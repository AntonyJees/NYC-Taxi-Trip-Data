```python
%pip install pyarrow==19.0.1
import urllib.request
from pathlib import Path
import pyarrow.parquet as pq
path=Path("yellow_tripdata_2025-01.parquet")
if not path.exists():
    urllib.request.urlretrieve("https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2025-01.parquet",path)
file=pq.ParquetFile(path)
print("Rows:",file.metadata.num_rows)
assert file.metadata.num_rows >= 1_000_000
print(file.schema_arrow)
sample=next(file.iter_batches(batch_size=1000)).to_pandas()
sample.head()
```
